```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = 0
b = 1 
N = 400
h = (b-a)/N
```

```python
def f(x,y):
    return np.exp(-x*y)

def Simpson1(h,y):
    x = np.arange(a,b,2*h)
    return (h/3)*np.sum(f(x,y)+4*f(x+h,y)+f(x+2*h,y))

print(Simpson1(h,1))
```

```python
valores_y = np.linspace(a, b, N+1)
F = np.zeros(N+1)
print(valores_y.shape, F.shape)
```

```python
for j in range(N+1):
    F[j] = Simpson1(h, valores_y[j])
print(F)
```
