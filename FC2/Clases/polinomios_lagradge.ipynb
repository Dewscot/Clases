```python
import numpy as np
```

```python


x = np.array([0., 1., 2.])
y = np.array([1., 2., 5.])

# Construcción explícita de Lagrange
def lagrange(z, x, y):
    resultado = 0.0

    for i in range(len(x)):
        L = 1.0

        for j in range(len(x)):
            if j != i:
                L *= (z - x[j]) / (x[i] - x[j])

        resultado += y[i] * L

    return resultado

# Comparación con polyfit
coef = np.polyfit(x, y, 2)

print("Lagrange:", lagrange(0.5, x, y))
print("polyfit:", np.polyval(coef, 0.5))
print("Coeficientes:", coef)
```
