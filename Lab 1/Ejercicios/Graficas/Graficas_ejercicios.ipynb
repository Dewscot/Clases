# Práctica — cinco tipos de gráficos (guía de clase 1)
**Objetivo:** reconocer qué muestra cada representación. Datos: los cinco archivos de la carpeta.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('bmh')
```

## 1. Histograma — frecuencia de tiempos
Cada barra cuenta estudiantes dentro de un intervalo de **5 min**.

```python
datos = np.genfromtxt('01_histograma.txt')
tiempos = datos[:, 1]
plt.hist(tiempos, bins=np.arange(30, 61, 5), edgecolor='white')
plt.xlabel('Tiempo en completar la prueba (min)')
plt.ylabel('Frecuencia (estudiantes)')
plt.title('Histograma de tiempos')
plt.show()
```

## 2. Barras — comparación de categorías

```python
datos = np.genfromtxt('02_barras.txt', dtype=str)
animales, cantidad = datos[:, 0], datos[:, 1].astype(int)
plt.bar(animales, cantidad)
plt.xlabel('Tipo de animal')
plt.ylabel('Cantidad vendida')
plt.show()
```

## 3. Circular — partes de un total

```python
datos = np.genfromtxt('03_circular.txt', dtype=str)
categorias = np.char.replace(datos[:, 0], '_', ' ')
gastos = datos[:, 1].astype(float)
assert gastos.sum() == 500
plt.pie(gastos, labels=categorias, autopct='%1.0f%%')
plt.title('Distribución de gastos')
plt.show()
```

## 4. XY lineal — crecimiento de la planta

```python
datos = np.genfromtxt('04_xy_lineal.txt')
dia, altura = datos.T
plt.scatter(dia, altura)
plt.xlabel('Tiempo (días)')
plt.ylabel('Altura (cm)')
plt.show()
```

## 5. XY semilogarítmico — crecimiento exponencial
Si la cantidad se duplica por unidad de tiempo, $N(t)=N_0\,2^t$. Basta el eje vertical logarítmico; **no** se transforma el tiempo porque aquí aparece $t=0$.

```python
datos = np.genfromtxt('05_xy_logaritmico.txt')
hora, bacterias = datos.T
plt.scatter(hora, bacterias)
plt.yscale('log')
plt.xlabel('Tiempo (h)')
plt.ylabel('Número de bacterias')
plt.show()
```
