```python
import numpy as np
import matplotlib.pyplot as plt 
datos = np.genfromtxt('Temperatura-vs-tiempo_CON_INCERTIDUMBRES.txt', skip_header=3)
escala = np.linspace(0,16,160)
t = datos[:,0] # en segundo
delta_t = datos[:,1]
T = datos[:,2]# en grados
delta_T = datos[:,3]
#print(len(t))
#print(escala)
plt.scatter(t,T,)
plt.errorbar(t, T, xerr=delta_t, yerr=delta_T, fmt='.')
plt.xlabel('$Tiempo$')
plt.ylabel('$Temperatura$')
plt.grid('true')
plt.savefig('grafico-lineal-lineal')
```

```python
plt.scatter(t,T,)
plt.errorbar(t, T, xerr=delta_t, yerr=delta_T, fmt='.')
plt.xlabel('$Tiempo$')
plt.ylabel('$Temperatura$')
plt.yscale("log")
plt.grid('true')
plt.savefig('grafico-lineal-lineal')
```

```python
def logaritmica(T):
    return np.log(T/T[0])
```

```python
#print(logaritmica(T)
len(logaritmica(T))
```

```python
plt.scatter(t,logaritmica(T))
plt.savefig('funcion-logaritmica-lineal-lineal')
```

```python
m,b = np.polyfit(t,logaritmica(T),1)
print(m,b)
print(logaritmica(T)[0])
ajuste_lineal = np.polyval([m,b], t)
plt.plot(t,ajuste_lineal)
plt.scatter(t,logaritmica(T))
```
