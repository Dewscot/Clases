# Gráficos — clase de Laboratorio I
**Objetivo:** elegir la representación correcta y mostrar incertidumbres. Conserva los métodos de la clase: `genfromtxt`, `errorbar`, `bar` e `histogram`.

## 1. Gráfico XY y barras de error
En cada punto medimos $t\pm\Delta t$ y $d\pm\Delta d$. El tiempo va en el eje horizontal porque es la variable independiente.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('bmh')

datos = np.genfromtxt('datos_distancia_versus_tiempo.txt')
d, Dd, t, Dt = datos.T
print(f'{len(t)} mediciones')
```

```python
plt.figure(figsize=(7, 4))
plt.errorbar(t, d, xerr=Dt, yerr=Dd, fmt='o', capsize=3)
plt.xlabel('Tiempo (s)')
plt.ylabel('Distancia (m)')
plt.title('Distancia en función del tiempo')
plt.tight_layout()
plt.show()
```

**Lectura:** el punto central es $(t_i,d_i)$; las barras indican $t_i\pm\Delta t_i$ y $d_i\pm\Delta d_i$. Si una barra casi no se ve, no significa incertidumbre cero.

## 2. Barras frente a XY
`bar` puede representar columnas, pero para una relación entre dos magnitudes continuas con incertidumbres resulta más informativo `errorbar`.

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 3.5))
ax[0].errorbar(t, d, xerr=Dt, yerr=Dd, fmt='o', capsize=2)
ax[0].set(title='Puntos e incertidumbre', xlabel='Tiempo (s)', ylabel='Distancia (m)')
ax[1].bar(t, d, width=0.05)
ax[1].set(title='Barras (comparación)', xlabel='Tiempo (s)', ylabel='Distancia (m)')
plt.tight_layout()
plt.show()
```

## 3. Histograma de una magnitud
El histograma cuenta valores por **intervalos**. Usamos una semilla fija para repetir el mismo ejemplo al reabrir el notebook.

```python
rng = np.random.default_rng(42)
x = 10 * rng.random(20)
conteo, bordes = np.histogram(x, bins=np.arange(0, 11))
print('Cantidad entre 1 y 2:', np.count_nonzero((x >= 1) & (x < 2)))
print('Frecuencias:', conteo)
plt.hist(x, bins=bordes, edgecolor='white')
plt.xlabel('Valor de x')
plt.ylabel('Frecuencia')
plt.title('Histograma con intervalos de ancho 1')
plt.show()
```
