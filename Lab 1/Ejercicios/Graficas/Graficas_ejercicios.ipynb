```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
datos_histograma = np.genfromtxt('01_histograma.txt')
x = datos_histograma[:,0]
y = datos_histograma[:,1]

print(np.min(y),np.max(y))
print(len(x)) 
```

```python
plt.hist(y,6)
plt.grid= True
plt.xlim(30,60)
plt.xlabel('tiempo')
plt.ylabel('cantidad de estudiantes')
plt.title("tiempo que demoran los estudiantes en completar la prueba")
plt.show()
```

```python
dato_barras1=np.genfromtxt('02_barras.txt',dtype='str')
dato_barras2=np.genfromtxt('02_barras.txt',dtype='int')
x= dato_barras1[:,0]
y= dato_barras2[:,1]
plt.bar(x,y)
plt.ylim(0,16)
plt.xlabel('animales vendidos')
plt.title('cantidad de animales vendidos')
```

```python
datos_circulo1 = np.genfromtxt('03_circular.txt',dtype='str')
datos_circulo2 = np.genfromtxt('03_circular.txt',dtype='int')

x= datos_circulo1[:,0]
y=datos_circulo2[:,1]
# print(x,y)
plt.pie(y,labels=list(x))
```

```python
datos_lineal = np.genfromtxt('04_xy_lineal.txt')
x= datos_lineal[:,0]
y= datos_lineal[:,1]

plt.scatter(x,y)
plt.xlabel("dia")
plt.ylabel("altura[cm]")
plt.title("Crecimiento de una planta de tomate")
```

```python
datos_logaritmo = np.genfromtxt('05_xy_logaritmico.txt')
x= datos_logaritmo[:,0]
y= datos_logaritmo[:,1]

plt.scatter(x,y)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Hora')
plt.ylabel('Numero de bacterias')
plt.title('Crecimiento de bacterias por hora')
```
