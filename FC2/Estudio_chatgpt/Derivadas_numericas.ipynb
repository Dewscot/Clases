```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3

def df_exacta(x):
    return 3*x**2

def df(x, h):
    return (f(x+h) - f(x))/h

def E_abs(x, h):
    return np.abs(df_exacta(x) - df(x, h))
```

```python
h = np.array([0.1, 0.01])

print("Derivadas numéricas:", df(1, h))
print("Errores absolutos:", E_abs(1, h))
```

```python
h = np.array([1e-1, 1e-4, 1e-8, 1e-12, 1e-16, 1e-20])

for paso in h:
    print(f"h = {paso:.0e}, Error = {E_abs(1, paso):.8e}")
```

Mira como cuando estamos trabajando en h de 1e-12 su error es del mismo que el de 1e-04 por lo que cuanto más pequeño el h disminuye el error hasta ciertos puntos hasta que por redondear nos puede suceder que df se vuelva 0 como en h=1e-16 y más adelante 

```python
print(1.0 + 1e-16 == 1.0)

print(f(1.0 + 1e-16) - f(1.0))

print(df(1.0, 1e-16))
```
