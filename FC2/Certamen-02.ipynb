```python
import numpy as np
import matplotlib.pyplot as plt
```

Primer ejercicio

```python
#Definimos la funcio de dentro de la integral
def f (x,a):
    return (x*np.sin(x))/(1-2*a*np.cos(x)+a**2)
```

```python
#Definimos los intervalos de la integral
inicio = 0
final = np.pi

N = 100  #Los intervalos para el a 
h = 0.01 #deberia de ser 0.01, no lo es
constante = np.linspace(0,2,N+1) #intervalo de 0 a 2
x = np.linspace(inicio,final,N+1)
print(constante.shape, x.shape)
```

```python
integral = np.zeros(len(constante))
suma = 0
for i  in range (N+1):
    a = constante[i]
    integral[i] = (h/2)*np.sum(f(x,a)+f(x+h,a))

nombres = "a", "I(a)"
plt.plot(constante,integral)
plt.xlabel(nombres[0])
plt.ylabel(nombres[1])
```

b) calculamos la funcion exacta para ver su valor

```python
def I_exact (a):
    if 0<a<1:
        return (np.pi/a)*np.log(1+a)
    else:
        return (np.pi/a)*np.log(1+1/a)
```

```python
Error = np.zeros(N+1)
for i  in range (N+1):
    a = constante[i]
    I_error = integral[i]-I_exact(a)
    Error[i] = np.abs(I_error)
print(Error)
```

```python
plt.plot(integral,Error)
```
