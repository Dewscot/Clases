# Integración: rectángulo, punto medio y trapecio

N cuenta subintervalos, no nodos. Las sumas aproximan la integral con signo.

**Uso:** ejecuta los imports y el bloque del caso que necesites. Cada ejemplo empieza con sus propios datos. `def` se usa solo para expresiones matemáticas.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Métodos básicos

Cambia f, a, b y N. Rectángulo izquierdo: altura inicial; punto medio: altura central; trapecio: promedio de los extremos. Se muestran los tres con la misma malla.

```python
def f(x):
    return x**3

a, b = 1., 2.
N = 100
assert N > 0 and isinstance(N, int)
h = (b-a)/N
x = np.linspace(a, b, N+1)
# Cada arreglo tiene N contribuciones
I_rect = h*np.sum(f(x[:-1]))
I_medio = h*np.sum(f((x[:-1]+x[1:])/2))
I_trap = h/2*np.sum(f(x[:-1])+f(x[1:]))
print('Rectángulo:', I_rect, 'Punto medio:', I_medio, 'Trapecio:', I_trap)
print('Referencia:', (b**4-a**4)/4)
```

## 2. Trapecio con ciclo explícito

Alternativa equivalente a la suma vectorizada. El último i es N−1; i+1 llega a N.

```python
def f(x):
    return x**3
a, b, N = 1., 2., 100
h = (b-a)/N
x = np.linspace(a,b,N+1)
suma = 0.
for i in range(N):
    suma += f(x[i]) + f(x[i+1])
I = h/2*suma
print(I)
assert np.isclose(I, h/2*np.sum(f(x[:-1])+f(x[1:])))
```

## 3. Datos discretos, incluso no uniformes

Cada trapecio tiene su propio ancho. No asumir h constante. Los puntos deben estar ordenados en el sentido de integración.

```python
x = np.array([0., .2, .7, 1.])
y = 1+2*x
assert len(x)==len(y) and len(x)>=2
assert np.all(np.diff(x)>0) or np.all(np.diff(x)<0)
I = 0.
for i in range(len(x)-1):
    h_i = x[i+1]-x[i]
    I += h_i*(y[i]+y[i+1])/2
print(I)
assert np.isclose(I,2.)
```

## 4. Comprobar convergencia

Para funciones suaves: rectángulo izquierdo O(h), trapecio y punto medio O(h²). Refinar N y comprobar el resultado.

```python
def f(x):
    return np.exp(x)
a,b=0.,1.
for N in [10,20,40,80]:
    h=(b-a)/N
    x=np.linspace(a,b,N+1)
    I=h/2*np.sum(f(x[:-1])+f(x[1:]))
    print(N,I,abs(I-(np.e-1)))
```
