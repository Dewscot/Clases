```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
intervalo_inicial = 1
intervalo_final = 2
a = intervalo_inicial
b = intervalo_final
N = 100 #cantidad de intervalos
h =  (b-a)/N
```

```python
def f(x):
    return x**3
```

```python
x = np.arange(a,b,h)
plt.bar(x,f(x))
```

```python
#np.arange(a,b,h)
```

Primer version para calcular la integral

```python
def rectangulo(h):
    x = np.arange(a,b,h)
    s=0
    for x_i in x:
        s = s +f(x_i)*h
    return s
```

```python
rectangulo(h)
```

Segunda forma, con numpy

```python
def rectangulo2(h):
    x = np.arange(a,b,h)
    return h*np.sum(f(x))
rectangulo2v = np.vectorize(rectangulo2) #esto para poder varios valores de h, (00.1,0.1)
```

```python
rectangulo2(h)
```

```python
#Comparacion
%time rectangulo(h)
%time rectangulo2(h)
rectangulo2(h)-rectangulo(h)

#Es mejor hacerlo con numpy que con ciclo for
```

### Regla trapezoidal

```python
def g2(h):
    x = np.arange(a,b,h)
    plt.bar(x,f(x))
    plt.plot(x,f(x))
```

```python
g2(h)
```

```python
def trapezoide(h):
    x = np.arange(a,b,h)
    return np.sum(f(x)+f(x+h))*h/2
trapezoidev = np.vectorize(trapezoide)
```

```python
trapezoide(h)
```
