```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('bmh')
```

```python
datos = np.genfromtxt('datos_distancia_versus_tiempo.txt')
d = datos[:,0]
Dd = datos[:,1]
t = datos[:,2]
Dt = datos[:,3]

```

```python
plt.errorbar(t,d, xerr=Dt , yerr=Dd, fmt='o')
plt.xlabel("$t$,(s)")
plt.ylabel("$d$,(m)")
```

```python
plt.bar(t,d, width=0.1)
```

```python
plt.figure(figsize=[10,4])
plt.subplot(1,1,1)
plt.errorbar(t,d, xerr=Dt , yerr=Dd, fmt='o')
plt.xlabel("$t$,(s)")
plt.ylabel("$d$,(m)")
plt.subplot(1,2,2)
plt.bar(t,d, width=0.01)

```
