## 1. Media y mediana
Una observación extrema puede mover mucho la media y poco la mediana.

```python
x = np.array([9.8, 9.9, 10.0, 10.1, 12.2])
print('Media:', np.mean(x), 'cm')
print('Mediana:', np.median(x), 'cm')
```

## 2. Varianza y desviación estándar
Usamos el denominador $N$, como en la presentación: $\sigma=\sqrt{\frac1N\sum_i(x_i-\bar x)^2}$.

```python
x = np.array([7., 10., 13.])
varianza = np.sum((x - np.mean(x))**2) / len(x)
sigma = np.sqrt(varianza)
print(f'Varianza: {varianza:.2f} cm²; desviación: {sigma:.2f} cm')
```

## 3. Error estándar del promedio
La división es por $\sqrt{N}$, **no por $N$**. Por ejemplo, al pasar de cuatro a dieciséis mediciones, el error estándar baja a la mitad si $\sigma$ se mantiene.

```python
sigma = 2.4
for N in (4, 16):
    print(f'N={N}: δx̄ = {sigma / np.sqrt(N):.2f} cm')
```

## 4. Propagación: área $A=ab$
Lineal: $\delta A=|b|\delta a+|a|\delta b$. Cuadrática (independientes): $\delta A=\sqrt{(b\delta a)^2+(a\delta b)^2}$.

```python
a, da = 2., 0.1
b, db = 3., 0.2
A = a*b
dA_lineal = abs(b)*da + abs(a)*db
dA_cuadratica = np.sqrt((b*da)**2 + (a*db)**2)
print(f'Área: {A:.1f} m²; incertidumbres: {dA_lineal:.1f} y {dA_cuadratica:.1f} m²')
```
