
## Método de bisección

Permite encontrar un cero de una función continua $f(x)$ dentro de un intervalo $[a,b]$, siempre que exista un cambio de signo:

$$
f(a)f(b)<0
$$

En cada iteración calculamos el punto medio:

$$
c=\frac{a+b}{2}
$$

- Si $f(a)f(c)<0$, actualizamos $b=c$.
- En caso contrario, actualizamos $a=c$ (si $f(c)\neq 0$).

Repetimos hasta alcanzar la tolerancia deseada:

$$
\epsilon=\frac{b-a}{2}<\epsilon_{\mathrm{tol}}
$$

Es un método robusto, pero de convergencia relativamente lenta.


Ejemplo

```python
def f(x):
    return x*np.sin(x)-1

a, b = 0, 2
tolerancia = 1e-6
error = 0.5*(b-a)

assert f(a)*f(b) < 0, "No hay cambio de signo"

while error > tolerancia:
    c = 0.5*(a+b)

    if abs(f(c)) < tolerancia:
        break

    if f(a)*f(c) < 0:
        b = c
    else:
        a = c

    error = 0.5*(b-a)

print(c)
```


**Limitación:** Bisección necesita un cambio de signo para garantizar una raíz. En $g(x)=x^2$ existe una raíz en $x=0$, pero la función no cambia de signo, por lo que el método no garantiza encontrarla.



## Método de Newton-Raphson

Permite encontrar una raíz de $f(x)=0$ utilizando su derivada y un valor inicial (semilla).

$$
\boxed{x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}}
$$

Converge rápidamente cerca de una raíz simple, pero depende de la semilla elegida y requiere conocer $f'(x)$, que no debe anularse durante las iteraciones.


```python
tolerancia, stop = 1e-6, False
x = 1.0  # Semilla inicial

def df(x):
    return np.sin(x) + x*np.cos(x)

while not stop:
    h = -f(x)/df(x)
    x = x+h
    stop = abs(h)<tolerancia or abs(f(x))<tolerancia

print(x)
```

Puede ir hacia otros ceros y si esta en la mitad de 2 ceros, suceden problemas, aunque es màs eficiente, todo depende de la semilla (un solo punto para calcular la derivada), cuando màs cerca mejor. Require saber la derivada analitica

```python
tolerancia, stop = 1e-6, False
x = 1.0 # valor semilla
def dg(x):   
    return 2*x 

while(not stop):
    h = -g(x)/dg(x)
    x = x+h
    stop = abs(h)<tolerancia or abs(g(x))<tolerancia
print(x)    
```


**Observación:** Newton-Raphson puede encontrar raíces sin cambio de signo. Para $g(x)=x^2$, la convergencia es lineal porque la raíz tiene multiplicidad 2. Además, $|g(x)|<tol$ no garantiza que el error en $x$ sea menor que $tol$.



## Método de la secante

Similar a Newton-Raphson, pero aproxima la derivada utilizando dos puntos anteriores:

$$
\boxed{x_{n+1}=x_n-f(x_n)\frac{x_n-x_{n-1}}{f(x_n)-f(x_{n-1})}}
$$

No requiere conocer la derivada analítica, pero necesita dos semillas iniciales. Su convergencia depende de las semillas elegidas.


```python
tolerancia, stop = 1e-6, False
x0, x1 = 1.0, 1.5  # Dos semillas
i = 0

while not stop:
    i += 1
    h = -f(x1)*(x1-x0)/(f(x1)-f(x0))
    x0, x1 = x1, x1+h
    stop = abs(h)<tolerancia or abs(f(x1))<tolerancia

print(x1)
print(i)
```

Requiere 2 puntos (no necesariamente cerca de la raiz) para poder calcular su derivada y realizamos una secante con ello, de ahì sacamos 2 nuevos puntos y se continua. ventajas, solo necesitamos el valor de la funcion en los 2 puntos, no f'
