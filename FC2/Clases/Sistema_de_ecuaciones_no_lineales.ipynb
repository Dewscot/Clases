## Método de Newton-Raphson

Permite encontrar una raíz de $f(x)=0$ utilizando su derivada y un valor inicial (semilla).

$$
\boxed{x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}}
$$

Vamos a trabajar con este metodo

## Ejemplo:
$f_0(x,y)=y-\cos(x)=0$ y $f_1(x,y)= x-e^{y-1}=0$

Primero siempre es bueno ver la grafica

```python
x = np.linspace(0,1,100)
y = np.linspace(0,1,100)
plt.plot(x,np.cos(x))
plt.plot(np.exp(x-1),x)
plt.show()
```

```python
def f(x):
    return np.array([x[1]-np.cos(x[0]), x[0]-np.exp(x[1]-1)])
def J(x):
    return np.array([
                    [np.sin(x[0]),1],
                    [1,np.exp(x[1]-1)]
                    ])
```

```python
x0= np.array([1,1])
f(x0)
J(x0)
```

Ocupamos esta funcion para determinar la inversa 

```python
np.linalg.inv(J(x0))
#np.linalg.inv(J(x0)) @ J(x0)
```

```python
Dx = -np.linalg.inv(J(x0)) @ f(x0)
print(Dx)
print(x0+Dx)
```

```python
np.linalg.solve(J(x0),-f(x0))
```

```python
x0 = np.array([1,1])
for i in range (10):
    print(x0)
    Dx = np.linalg.solve(J(x0),-f(x0))
    x0 = x0+Dx
```
