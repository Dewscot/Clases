```python
def derivada_adelantada(f, x, h):
    return (f(x+h) - f(x))/h

def derivada_atrasada(f, x, h):
    return (f(x) - f(x-h))/h

def derivada_centrada(f, x, h):
    return (f(x+h) - f(x-h))/(2*h)

def segunda_derivada(f, x, h):
    return (f(x+h) - 2*f(x) + f(x-h))/h**2
```

```python
def f(x):
    return np.sin(x)

x, h = 1.0, 0.01

print("Adelantada:", derivada_adelantada(f, x, h))
print("Centrada:", derivada_centrada(f, x, h))
print("Exacta:", np.cos(x))
```
