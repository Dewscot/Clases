```python

Npuntos = 400
t_i = 0.0
t_f = 40.0

tiempos = np.linspace(t_i, t_f, Npuntos)
dt = tiempos[1] - tiempos[0]

trayectoria = np.zeros((Npuntos, 2))

trayectoria[0] = np.array([0.1,0])

for n in range(Npuntos-1):
    trayectoria[n+1]=trayectoria[n] + dt*np.array([trayectoria[n,1],-k/m*trayectoria[n,0]-g])

datos_y = trayectoria[:,0]

plt.plot(tiempos,datos_y)
print(trayectoria[1,0])
```

```python
omega = np.sqrt(k/m)
y0 = trayectoria[0, 0]

y_analitica = (y0+g/omega**2)*np.cos(omega*tiempos)-g/omega**2

plt.plot(tiempos,datos_y)
plt.plot(tiempos,y_analitica)
print(y_analitica[0])
print(trayectoria[0, 0])
plt.xlim(0,5)
plt.ylim(-10,10)
```

```python
Energia_euler = np.zeros(Npuntos)
for n in range(Npuntos):
    Energia_euler[n] = 1/2*m*trayectoria[n,1]**2+1/2*k*trayectoria[n,0]**2+m*g*trayectoria[n,0]
```

```python
len(Energia_euler)
print(Energia_euler[0])
```

```python
plt.plot(tiempos,Energia_euler)
```

```python
trayectoria_EC = np.zeros((Npuntos, 2))
trayectoria_EC[0] = np.array([0.1, 0.0])

trayectoria_EC[1,1] = trayectoria_EC[0,1]+dt*(-k/m*trayectoria_EC[0,0]-g)

trayectoria_EC[1,0] = trayectoria_EC[0,0]+dt*trayectoria_EC[1,1]

for n in range(Npuntos-1):
    trayectoria_EC[n+1,1] = trayectoria_EC[n,1]+dt*(-k/m*trayectoria_EC[n,0]-g)
    trayectoria_EC[n+1,0] = trayectoria_EC[n,0]+dt*trayectoria_EC[n+1,1]

plt.plot(tiempos,trayectoria_EC[:,0])
```

```python
Energia_euler_cromer = np.zeros(Npuntos)
for n in range(Npuntos):
    Energia_euler_cromer[n] = 1/2*m*trayectoria_EC[n,1]**2+1/2*k*trayectoria_EC[n,0]**2+m*g*trayectoria_EC[n,0]
plt.plot(tiempos,Energia_euler_cromer)
#plt.plot(tiempos,Energia_euler)
```

```python
trayectoria_SR = np.zeros((Npuntos,2))

trayectoria_SR[0] = np.array([0.1, 0.0])
trayectoria_SR[0,1] = dt/2*(-k/m*trayectoria_SR[0,0]-g)
    
for n in range (Npuntos-1):

    trayectoria_SR[n+1,0] = trayectoria_SR[n,0]+dt*trayectoria_SR[n,1] 

    trayectoria_SR[n+1,1] = trayectoria_SR[n,1]+dt*(-k/m*trayectoria_SR[n+1,0]-g)

plt.plot(tiempos, trayectoria_SR[:,0], label="Salto de rana")
plt.plot(tiempos, y_analitica, label="Analítica")
plt.legend()
plt.show()
```

```python
v1 = (trayectoria_SR[1:,1]+trayectoria_SR[:-1,1])/2
velocidades_SR =np.concatenate(([0.0], v1))
```

```python
Energia_SR = np.zeros(Npuntos)
for n in range(Npuntos):
    Energia_SR[n] = 1/2*m*velocidades_SR[n]**2+1/2*k*trayectoria_SR[n,0]**2+m*g*trayectoria_SR[n,0]
plt.plot(tiempos,Energia_SR)
```

```python
trayectoria_RK2 = np.zeros((Npuntos, 2))
trayectoria_RK2[0] = np.array([0.1, 0.0])
```

```python
for n in range (Npuntos-1):
    actual = trayectoria_RK2[n]
    derivada = np.array([actual[1],-k/m*actual[0]-g])
    intermedio = actual + dt/2*derivada
    derivada_intermedia = np.array([intermedio[1],-k/m*intermedio[0]-g])
    trayectoria_RK2[n+1]= actual+dt*derivada_intermedia
```

```python
plt.plot(tiempos,trayectoria_RK2[:,0])
plt.plot(tiempos,y_analitica)
```

```python
Energia_RK2 = np.zeros(Npuntos)
for n in range(Npuntos):
    Energia_RK2[n]= 1/2*m*trayectoria_RK2[n,1]**2+1/2*k*trayectoria_RK2[n,0]**2+m*g*trayectoria_RK2[n,0]
```

```python
plt.plot(tiempos,Energia_RK2)
print(Energia_RK2[0])
print(Energia_RK2[-1])
```

```python
trayectoria_RK4 = np.zeros((Npuntos, 2))
trayectoria_RK4[0] = np.array([0.1, 0.0])
```

```python
for n in range(Npuntos-1):
    actual = trayectoria_RK4[n]
    pendiente1 = np.array([actual[1],-k/m*actual[0]-g])
    medio2 = actual + dt/2*pendiente1
    pendiente2 = np.array([medio2[1],-k/m*medio2[0]-g]) 
    medio3 = actual + dt/2*pendiente2
    pendiente3 = np.array([medio3[1],-k/m*medio3[0]-g])
    medio4 = actual+ dt*pendiente3
    pendiente4 = np.array([medio4[1],-k/m*medio4[0]-g])
    trayectoria_RK4[n+1]= actual+dt/6*(pendiente1+2*pendiente2+2*pendiente3+pendiente4)
```

```python
plt.plot(tiempos,trayectoria_RK4[:,0])
plt.plot(tiempos,y_analitica)
```

```python
Energia_RK4 = np.zeros(Npuntos)
for n in range(Npuntos):
    Energia_RK4[n]= 1/2*m*trayectoria_RK4[n,1]**2+1/2*k*trayectoria_RK4[n,0]**2+m*g*trayectoria_RK4[n,0]
```

```python
plt.plot(tiempos,Energia_RK4)
print(Energia_RK4[0])
print(Energia_RK4[-1])
```
