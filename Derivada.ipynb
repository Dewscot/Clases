```python
import numpy as np 
import matplotlib.pyplot as plt
```

Las variables son x y t que son generadas del archi

```python

def f(x):
    return x**3
def fp(x) : 
    return 3*x**2
def f_num(x,h):
    return (f(x+h)-f(x))/h
def E_abs(x,h):
    return np.abs(f_num(x,h)-(fp(x+h)-fp(x)/h))
```

```python
print(f_num(1,0.1))
```

```python
print(E_abs(1,0.1))
```
