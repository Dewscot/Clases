```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
inicio_1 = 0
final_1 = 5

M = 100
N = 100000

valores_x1 = np.linspace(inicio_1, final_1, M+1)

C = np.zeros(M+1)
S = np.zeros(M+1)

def Cx(t):
    return np.cos(np.pi*t**2/2)

def Sx(t):
    return np.sin(np.pi*t**2/2)

for j in range(M+1):

    x = valores_x1[j]
    h = (x-inicio_1)/N

    t = inicio_1 + np.arange(0,N,2)*h

    C[j] = (h/3)*np.sum(Cx(t)+4*Cx(t+h)+Cx(t+2*h))
    S[j] = (h/3)*np.sum(Sx(t)+4*Sx(t+h)+Sx(t+2*h))


#print(C,S)
```

```python
print(C.shape, S.shape)
print(C[0], S[0])
print(C[-1], S[-1])
```

```python
plt.plot(valores_x1, C, label="C(x)")
plt.plot(valores_x1, S, label="S(x)")
plt.legend()
plt.show()
```

```python
from scipy.special import fresnel
```

```python
S_scipy, C_scipy = fresnel(valores_x1)
```

```python
print(C[-1], C_scipy[-1])
print(S[-1], S_scipy[-1])
```

```python
# N=400
Error_C = C_scipy-C
Error_S = S_scipy-S

print(Error_C[-1],Error_S[-1])
```

```python
error_max_C = np.max(np.abs(Error_C))
error_max_S = np.max(np.abs(Error_S))
```

```python
print(error_max_C,error_max_S)
```
