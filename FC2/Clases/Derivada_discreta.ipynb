```python
import numpy as np 
import matplotlib as plt
```

```python

x = np.array([1, 3, 3.2, 5.5])
f = np.array([-1, 7, 0, 3])

# Derivada discreta usando un for
N = len(x)
df = np.zeros(N)

for i in range(1, N-1):
    df[i] = (f[i+1] - f[i-1]) / (x[i+1] - x[i-1])

print(df)

# Versión vectorizada con NumPy
df_v = (f[2:] - f[:-2]) / (x[2:] - x[:-2])

print(df_v)
```
