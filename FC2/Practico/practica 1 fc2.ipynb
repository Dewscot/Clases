# expresion analoga del el error absoluto de la derivada adelantada para la derivad centrada 

el error absouto de una derivada exacata respecto a su numerica se da segun la siguiente expresion 
$$E_{abs}=\left|f'(x)-\overline{\frac{f(x+h)-f(x-h)}{2h}}\right|$$
donde 
$$\overline{f(x+h)} = f(x+h)(1+e_0)$$
$$\overline{f(x-h)} = f(x+h)(1+e_+)$$
de esta forma 
$$E_{abs}=|f'(x)-\frac{f(x+h)-f(x-h)}{2h}-\frac{e_+f(x+h)-e_0f(x-h)}{2h}|$$
donde se $f'(x)$ se puede escribir segun su *serie de taylor* de primer orden de la siguiente forma 
$$f'(x)= \frac{f(x+h)-f(x-h)}{2h}-h^2\frac{f'''(x)}{6}$$
$$f'(x)-\frac{f(x+h)-f(x-h)}{2h} =-h^2\frac{f'''(x)}{6}$$
de esta forma podemos escribir el error absoluto de la siguiente forma
$$E_{abs}=|-h^2\frac{f'''(x)}{6}-\frac{e_+f(x+h)-e_0f(x-h)}{2h}|$$
luego podemos hacer si siguientes asunciones 
$$e_0=e_1=e$$
$$f(x+h)=f(x-h)$$
para h pequeño y asumiendo que error de truncamiento de $e_0$ tiene un error de redondeo similar a $e_+$
De esta forma podemos escribir el error absoluto como siguie

$$E_{abs} \thickapprox |-h^2\frac{f'''(x)}{6}-\frac{ef(x)-ef(x)}{2h}|$$
$$E_{abs} \thickapprox |-h^2\frac{f'''(x)}{6}-\frac{2ef(x)}{2h}|$$
$$E_{abs} \thickapprox |-h^2\frac{f'''(x)}{6}-\frac{ef(x)}{h}|$$
y usando la desigualdad de minkowski, se llega a 
$$E_{abs}  \leq |h^2\frac{f'''(x)}{6}| + |\frac{e}{h}f(x)|$$





```python
import numpy as np 
import matplotlib.pyplot as plt
plt.style.use('bmh')
def f(x):
    return x**3
def fp(x):
    return 3*x**2
def fp2(x):
    return 6*x
def fp3(x):
    return 6
def fp_def(x,h): #derivada por definicion
    return (f(x+h)-f(x-h))/(2*h)
def E_abs(x,e,h): # error causado por la serie se taylor
    return np.abs((h**2*fp3(x))/6) + np.abs(e*f(x)/h)
e = 10**-15
x = np.geomspace(1e-20,1e-1,40)
h = np.geomspace(1e-20,1e-1,40)
error = np.abs(fp_def(1,h)-fp(1))
plt.scatter(h, error)

plt.xscale('log')
plt.yscale('log')

plt.xlabel("h")
plt.ylabel("error absoluto")
plt.plot(h, E_abs(1,e,h))
```
