```python
datos = np.genfromtxt('millikan.txt')
x = datos[:,0]
y = datos[:,1]
print(x,y)
```

Veamos los graficos en grafico de xy

```python
plt.scatter(x,y)
plt.xlabel('frecuencia v [hz]')
plt.ylabel('diferencia de potencial V [Volts]')
```

```python
coef = np.polyfit(x,y,1)
print(coef)
```

Siendo $\frac{h}{e} $ el primer coeficiente y segundo el coeficiente de posicion

Comparando ambas graficas

```python
xx = np.linspace(min(x),max(x)*1.1,100)
def f(xx):
    return coef[0]*xx+coef[1]
```

```python
plt.scatter(x,y)
plt.scatter(xx,f(xx),alpha=0.2)
plt.xlabel('frecuencia v [hz]')
plt.ylabel('diferencia de potencial V [Volts]')
```

Mientras que si tomamos el valor de e para la carga del electron y ocupamos el coeficiente que determinamos con MMC $coef[0]=h/e$, para determinar el valor del coeficiente, podemos determinar h (constante de plack)

```python
e = 1.602e-19
h = coef[0]*e
print(h)
```
